```python
from sift_lib import SiftMatcher, SiftConfig, SiftMatcherType
import cv2

# discover features
sift = SiftMatcher()

img = SiftMatcher.load_image('set2/mod.png')
features, _ = sift.extract_features(img)
img_with_features = SiftMatcher.draw_features(img, features, rich_display=True)
SiftMatcher.show_image('SIFT - extract features', img_with_features)
```

```python
# extract features - regular display
img_with_features = SiftMatcher.draw_features(img, features, rich_display=False)
SiftMatcher.show_image('SIFT - extract features, regular display', img_with_features)
```

```python
# find matches between two images
# config: BFMatcher, Lowe=0.75
sift = SiftMatcher()

img1 = SiftMatcher.load_image('set2/ref.png')
img2 = SiftMatcher.load_image('set2/mod.png')
f1, d1 = sift.extract_features(img1)
f2, d2 = sift.extract_features(img2)

matches = sift.find_matches(d1, d2)
img_with_matches = SiftMatcher.draw_matches(img1, f1, img2, f2, matches[:1000])
SiftMatcher.show_image("BFMatcher, Lowe=0.75 - matching test", img_with_matches)
```

```python
# find matches, BFMatcher with cross check, L1 norm
config = SiftConfig(
    cross_check=True,
    norm_type=cv2.NORM_L1
)

sift = SiftMatcher(config)

img1 = SiftMatcher.load_image('set2/ref.png')
img2 = SiftMatcher.load_image('set2/mod.png')
f1, d1 = sift.extract_features(img1)
f2, d2 = sift.extract_features(img2)

matches = sift.find_matches(d1, d2)
img_with_matches = SiftMatcher.draw_matches(img1, f1, img2, f2, matches[:1000])
SiftMatcher.show_image("BFMatcher, cross check - matching test", img_with_matches)
```

```python
# find matches, FLANN matcher
config = SiftConfig(
    matcher_type=SiftMatcherType.FLANN
)

sift = SiftMatcher(config)

img1 = SiftMatcher.load_image('set2/ref.png')
img2 = SiftMatcher.load_image('set2/mod.png')
f1, d1 = sift.extract_features(img1)
f2, d2 = sift.extract_features(img2)

matches = sift.find_matches(d1, d2)
img_with_matches = SiftMatcher.draw_matches(img1, f1, img2, f2, matches[:1000])
SiftMatcher.show_image("FLANN matcher - matching test", img_with_matches)
```
